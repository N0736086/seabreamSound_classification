```python
import numpy as np
import wave
from mcap.reader import make_reader
from mcap_ros2.decoder import Decoder

def recover_hydrophones(mcap_path, tank_prefix="/tank1"):
    # Target topics based on your previous info
    targets = [f"{tank_prefix}/hydrophone1/audio", f"{tank_prefix}/hydrophone2/audio"]
    audio_data = {t: [] for t in targets}
    
    print(f"Opening {mcap_path} in manual stream mode...")
    
    with open(mcap_path, "rb") as f:
        reader = make_reader(f)
        # Initialize decoder without arguments
        decoder = Decoder()
        
        try:
            # We pass the reader's records directly to the decoder
            for topic, msg, msg_obj in decoder.decode(reader.iter_messages()):
                if topic in targets:
                    # Convert to numpy and handle the 1025-element metadata quirk
                    chunk = np.array(msg_obj.data, dtype=np.float32)
                    
                    if len(chunk) == 1025:
                        chunk = chunk[:1024]
                    
                    audio_data[topic].append(chunk)
        except Exception as e:
            # This is expected if the file is truncated or has the 'opcode 164' error at the end
            print(f"\nStopped reading at corruption point: {e}")

    # Process and save any data we managed to pull out
    for topic, segments in audio_data.items():
        if not segments:
            print(f"No data found for {topic}")
            continue
            
        print(f"Concatenating {len(segments)} segments for {topic}...")
        full_signal = np.concatenate(segments)
        
        # Remove DC Offset
        full_signal -= np.mean(full_signal)
        
        # Normalize
        max_val = np.max(np.abs(full_signal))
        if max_val > 0:
            full_signal /= max_val
            
        output_name = f"{topic.replace('/', '_').strip('_')}.wav"
        
        with wave.open(output_name, "wb") as wav_file:
            wav_file.setnchannels(1)     # Mono
            wav_file.setsampwidth(4)     # 32-bit (4 bytes)
            wav_file.setframerate(48000)
            wav_file.writeframes(full_signal.astype(np.float32).tobytes())
        print(f"Successfully saved: {output_name}")

if __name__ == "__main__":
    # Pointing to your specific file
    FILE = "/home/feliciano/Downloads/tank1_av_20260327_013504_022_0.mcap"
    recover_hydrophones(FILE)
```
