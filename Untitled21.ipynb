```python
import os
import zipfile
import subprocess
from pathlib import Path
from pdf2docx import Converter

def unzip_overleaf(zip_path, extract_to="latex_project"):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_to)
    return Path(extract_to)

def find_main_tex(project_path):
    # Look for main.tex or the first .tex file
    candidates = list(project_path.glob("*.tex"))
    for c in candidates:
        if c.name.lower() in ["main.tex", "thesis.tex", "paper.tex"]:
            return c
    return candidates[0] if candidates else None

def compile_latex(tex_file):
    cwd = tex_file.parent
    subprocess.run(["pdflatex", "-interaction=nonstopmode", tex_file.name], cwd=cwd)
    subprocess.run(["pdflatex", "-interaction=nonstopmode", tex_file.name], cwd=cwd)
    return tex_file.with_suffix(".pdf")

def convert_pdf_to_docx(pdf_path, output_docx="output.docx"):
    cv = Converter(str(pdf_path))
    cv.convert(output_docx, start=0, end=None)
    cv.close()
    return Path(output_docx)

def convert_latex_to_docx_pandoc(tex_file, output_docx="pandoc_output.docx"):
    subprocess.run([
        "pandoc",
        str(tex_file),
        "-s",
        "--from=latex",
        "--to=docx",
        "--output", output_docx,
        "--resource-path", str(tex_file.parent)
    ])
    return Path(output_docx)

def full_conversion_pipeline(zip_path):
    project = unzip_overleaf(zip_path)
    tex_file = find_main_tex(project)

    if tex_file is None:
        raise FileNotFoundError("No .tex file found in the project.")

    pdf = compile_latex(tex_file)

    # PDF → DOCX (best visual fidelity)
    docx_pdf = convert_pdf_to_docx(pdf, "converted_from_pdf.docx")

    # LaTeX → DOCX (best structural fidelity)
    docx_pandoc = convert_latex_to_docx_pandoc(tex_file, "converted_from_latex.docx")

    return {
        "pdf_based_docx": docx_pdf,
        "latex_based_docx": docx_pandoc,
        "pdf": pdf
    }

if __name__ == "__main__":
    results = full_conversion_pipeline("Aquaculture_anf_Fisheries_12_03_2026.zip")
    print("Generated files:", results)
```
