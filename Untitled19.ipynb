```python
from mcap.reader import make_reader
from mcap_ros2.reader import read_ros2_messages

def get_mcap_topics(file_path):
    """
    Extracts all unique topics and their message types from an MCAP file.
    """
    topics_found = {}

    with open(file_path, "rb") as f:
        reader = make_reader(f)
        
        # Access the summary to get channel information
        # Channels contain the topic name and the schema (message type)
        summary = reader.get_summary()
        
        if summary is None:
            print("No summary found. The file might be unindexed.")
            return None

        for channel_id, channel in summary.channels.items():
            topic_name = channel.topic
            # Get the schema name (e.g., 'std_msgs/msg/String')
            schema = summary.schemas.get(channel.schema_id)
            schema_name = schema.name if schema else "Unknown Type"
            
            topics_found[topic_name] = schema_name

    return topics_found

if __name__ == "__main__":
    # Replace with your actual file path
    mcap_file = "/home/feliciano/Downloads/tank2_av_20260327_002203_372_0.mcap"
    
    try:
        results = get_mcap_topics(mcap_file)
        
        if results:
            print(f"{'TOPIC':<50} | {'TYPE'}")
            print("-" * 80)
            for topic, msg_type in sorted(results.items()):
                print(f"{topic:<50} | {msg_type}")
    except FileNotFoundError:
        print(f"Error: File '{mcap_file}' not found.")
    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
import numpy as np
import wave
from mcap_ros2.reader import read_ros2_messages

def extract_separate_hydrophones(mcap_path, tank_prefix="/tank1"):
    # Target topics based on your provided info
    topics = [
        f"{tank_prefix}/hydrophone1/audio",
        f"{tank_prefix}/hydrophone2/audio"
    ]
    
    SAMPLE_RATE = 48000
    
    for topic in topics:
        print(f"Processing: {topic}")
        audio_segments = []
        
        with open(mcap_path, "rb") as f:
            reader = read_ros2_messages(f, topics=[topic])
            
            for msg in reader:
                chunk = np.array(msg.ros_msg.data, dtype=np.float32)
                
                # Metadata fix: Remove the 1025th element (status/header)
                if len(chunk) == 1025:
                    chunk = chunk[:1024]
                
                audio_segments.append(chunk)
        
        if not audio_segments:
            print(f"Skipping {topic}: No data found.")
            continue

        # Concatenate and Normalize
        full_signal = np.concatenate(audio_segments)
        full_signal -= np.mean(full_signal) # Remove DC Offset
        
        max_val = np.max(np.abs(full_signal))
        if max_val > 0:
            full_signal /= max_val

        # Create unique filename (e.g., tank1_hydrophone1.wav)
        clean_name = topic.replace("/", "_").strip("_")
        output_filename = f"{clean_name}.wav"

        print(f"Saving to {output_filename}...")
        with wave.open(output_filename, "wb") as wav_file:
            wav_file.setnchannels(1)      # Mono
            wav_file.setsampwidth(4)      # 32-bit
            wav_file.setframerate(SAMPLE_RATE)
            wav_file.writeframes(full_signal.astype(np.float32).tobytes())

    print("\nExtraction complete. All hydrophones saved as separate files.")

if __name__ == "__main__":
    FILE = "/home/feliciano/Downloads/tank1_av_20260327_013504_022_0.mcap"
    
    try:
        extract_separate_hydrophones(FILE)
    except Exception as e:
        print(f"Error: {e}")
```

```python
import numpy as np
import wave
import csv
import os
from mcap_ros2.reader import read_ros2_messages

def extract_with_timestamps(mcap_path, topic="/tank1/hydrophone1/audio"):
    SAMPLE_RATE = 48000
    audio_segments = []
    timing_data = [] # Stores (sample_index, ros_timestamp_ns)
    
    current_sample_count = 0
    
    print(f"Processing: {topic}")
    with open(mcap_path, "rb") as f:
        reader = read_ros2_messages(f, topics=[topic])
        
        for msg in reader:
            # 1. Get raw data and clean metadata (1025 -> 1024)
            chunk = np.array(msg.ros_msg.data, dtype=np.float32)
            if len(chunk) == 1025:
                chunk = chunk[:1024]
            
            # 2. Capture the ROS Log Time (Nanoseconds) for the start of this chunk
            # msg.log_time_ns is the precise time the message was recorded
            timing_data.append({
                'sample_index': current_sample_count,
                'ros_timestamp_ns': msg.log_time_ns
            })
            
            audio_segments.append(chunk)
            current_sample_count += len(chunk)

    if not audio_segments:
        print(f"No data for {topic}")
        return

    # 3. Save the WAV file
    full_signal = np.concatenate(audio_segments)
    base_name = topic.replace("/", "_").strip("_")
    wav_filename = f"{base_name}.wav"
    
    with wave.open(wav_filename, "wb") as wav_file:
        wav_file.setnchannels(1)
        wav_file.setsampwidth(4) # 32-bit Float
        wav_file.setframerate(SAMPLE_RATE)
        wav_file.writeframes(full_signal.astype(np.float32).tobytes())
    
    # 4. Save the Timestamp Map (Sidecar File)
    csv_filename = f"{base_name}_timestamps.csv"
    with open(csv_filename, 'w', newline='') as csvfile:
        fieldnames = ['sample_index', 'ros_timestamp_ns']
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(timing_data)

    print(f"Success!")
    print(f"Audio: {wav_filename}")
    print(f"Timestamps: {csv_filename} (Maps audio chunks to ROS time)")

if __name__ == "__main__":
    FILE = "/home/feliciano/Downloads/tank1_av_20260327_013504_022_0.mcap"
    
    # Extract for both hydrophones
    extract_with_timestamps(FILE, topic="/tank1/hydrophone1/audio")
    extract_with_timestamps(FILE, topic="/tank1/hydrophone2/audio")
```
