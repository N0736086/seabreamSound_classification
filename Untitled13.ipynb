```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import os

def generate_high_density_circular_image(file_path, save_path):
    # 1. Load 2s audio and downsample to 100 points
    y, sr = librosa.load(file_path, duration=2.0)
    
    indices = np.linspace(0, len(y) - 1, 100).astype(int)
    samples = y[indices]
    
    # 2. Max Normalization
    max_amp = np.max(np.abs(samples))
    if max_amp > 0:
        samples = samples / max_amp

    # 3. Create the Plot
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.axhline(0, color='black', linewidth=0.5, alpha=0.3)
    
    colors = plt.cm.viridis(np.linspace(0, 1, len(samples)-1))

    for i in range(len(samples) - 1):
        s = samples[i]
        e = samples[i+1]
        
        center = (s + e) / 2
        radius = abs(e - s) / 2
        
        if radius < 0.0001: continue 
        
        circle = plt.Circle((center, 0), radius, color=colors[i], fill=False, lw=0.7, alpha=0.6)
        ax.add_patch(circle)
        
        direction = 1 if e > s else -1
        for y_pos in [radius, -radius]:
            ax.annotate('', xy=(center + 0.01 * direction, y_pos), 
                        xytext=(center - 0.01 * direction, y_pos),
                        arrowprops=dict(arrowstyle='->', color=colors[i], lw=0.5, mutation_scale=8))

    # 4. Formatting
    ax.set_aspect('equal')
    plt.xlim(-1.1, 1.1)
    plt.ylim(-1.1, 1.1)
    plt.axis('off')
    
    # Corrected bg_color to facecolor
    plt.savefig(save_path, dpi=128, bbox_inches='tight', pad_inches=0, facecolor='white')
    plt.close()

# --- FIXED SECTION ---
# This should be the parent directory containing the class folders
main_audio_folder = "/home/feliciano/Downloads/Preliminary Data/AllFish_2secSplit"
output_image_folder = 'circular_dataset_100'

# Check if the directory exists
if not os.path.exists(main_audio_folder):
    print(f"Error: {main_audio_folder} is not a valid directory.")
else:
    for class_name in os.listdir(main_audio_folder):
        class_path = os.path.join(main_audio_folder, class_name)
        
        # Only process if it is a directory
        if os.path.isdir(class_path):
            save_class_path = os.path.join(output_image_folder, class_name)
            os.makedirs(save_class_path, exist_ok=True)
            
            for audio_file in os.listdir(class_path):
                if audio_file.endswith('.wav'):
                    generate_high_density_circular_image(
                        os.path.join(class_path, audio_file),
                        os.path.join(save_class_path, audio_file.replace('.wav', '.png'))
                    )
```
