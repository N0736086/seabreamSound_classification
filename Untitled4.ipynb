```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Setup directories
os.makedirs("data/images", exist_ok=True)

def generate_sequence(pattern_type):
    base = np.arange(1, 7)
    noise = np.random.uniform(-0.5, 0.5, 6)
    if pattern_type == "Ascending":
        return base + noise
    elif pattern_type == "Descending":
        return base[::-1] + noise
    elif pattern_type == "Oscillating":
        return np.array([1, 6, 1, 6, 1, 6]) + np.random.uniform(-0.8, 0.8, 6)
    else: # Random/Noise
        return np.random.uniform(1, 7, 6)

def save_circular_plot(seq, label, idx):
    norm_seq = seq / np.max(seq)
    fig, ax = plt.subplots(figsize=(2, 2)) # Small size for CNN input
    for i in range(len(norm_seq)-1):
        start, end = norm_seq[i], norm_seq[i+1]
        center, radius = (start + end)/2, abs(end - start)/2
        if radius < 0.01: continue
        circle = plt.Circle((center, 0), radius, color='black', fill=False, lw=1)
        ax.add_patch(circle)
        # Simplified arrows for small resolution
        flow_dir = 1 if end > start else -1
        ax.annotate('', xy=(center + 0.02*flow_dir, radius), xytext=(center - 0.02*flow_dir, radius),
                    arrowprops=dict(arrowstyle='->', lw=1, color='black'))
    
    ax.set_aspect('equal')
    plt.xlim(-0.1, 1.1); plt.ylim(-0.6, 0.6); plt.axis('off')
    plt.savefig(f"data/images/{label}_{idx}.png", dpi=50) # Low DPI is fine for CNN
    plt.close()

# Create Dataset
data_rows = []
classes = ["Ascending", "Descending", "Oscillating", "Random"]

for label_idx, label in enumerate(classes):
    for i in range(100):
        seq = generate_sequence(label)
        save_circular_plot(seq, label, i)
        data_rows.append(np.append(seq, label_idx))

df = pd.DataFrame(data_rows, columns=[f'v{i}' for i in range(6)] + ['label'])
```

```python
X = df.drop('label', axis=1)
y = df['label']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y)

rf = RandomForestClassifier(n_estimators=100)
rf.fit(X_train, y_train)
num_acc = accuracy_score(y_test, rf.predict(X_test))
print(f"Numerical Accuracy: {num_acc:.4f}")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix

# ==========================================
# 1. THE CIRCULAR TRANSFORMATION FUNCTION
# ==========================================
def transform_to_circles(X_input):
    # Convert input to float to avoid UFuncTypeError (string issues)
    X_numeric = np.array(X_input).astype(float)
    
    circular_data = []
    for seq in X_numeric:
        # Max-Normalization (x / max_val)
        max_val = np.max(seq)
        norm_seq = seq / max_val if max_val != 0 else seq
            
        # Extract Geometric Features based on your circular logic
        diffs = np.diff(norm_seq)
        diameters = np.abs(diffs)      # Width of the circles
        directions = np.sign(diffs)    # Convergent arrow direction
        centers = (norm_seq[:-1] + norm_seq[1:]) / 2  # Position on number line
        
        # Flatten into a feature vector: [diameters, directions, centers]
        feat_vector = np.hstack([diameters, directions, centers])
        circular_data.append(feat_vector)
        
    return np.array(circular_data)

# ==========================================
# 2. PROCESS AND SPLIT DATA (70/20/10)
# ==========================================
# Transform your raw X into the circular geometric representation
X_circular = transform_to_circles(X)
y_array = np.array(y)

# First split: 70% Training, 30% Temporary (Test + Val)
X_train, X_tmp, y_train, y_tmp = train_test_split(
    X_circular, y_array, test_size=0.30, stratify=y_array, random_state=42
)

# Second split: Split the 30% into Test (20%) and Val (10%)
# 1/3 of 30% is exactly 10%
X_test, X_val, y_test, y_val = train_test_split(
    X_tmp, y_tmp, test_size=1/3, stratify=y_tmp, random_state=42
)

# ==========================================
# 3. TRAIN CLASSIFIER
# ==========================================
# Pipeline ensures scaling is applied to test/val data correctly
clf = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(kernel='rbf', C=1.0, gamma='scale', probability=True))
])

clf.fit(X_train, y_train)

# ==========================================
# 4. EVALUATE PERFORMANCE
# ==========================================
y_pred = clf.predict(X_test)

print("--- Data Split Verification ---")
print(f"Total: {len(X_circular)} | Train: {len(X_train)} | Test: {len(X_test)} | Val: {len(X_val)}")
print("\n--- Circular Geometric Classification Report ---")
print(classification_report(y_test, y_pred))

# Verify accuracy on the unseen Validation set
val_accuracy = clf.score(X_val, y_val)
print(f"Final Validation Accuracy: {val_accuracy:.4f}")
```
