```python
import os
import librosa
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm

# Setup directories
data_path = '/home/feliciano/Downloads/Preliminary Data/AllFish_2secSplit'
output_img_path = 'dataset_images'
classes = [f for f in os.listdir(data_path) if os.path.isdir(os.path.join(data_path, f))]

def generate_circular_image(file_path, save_path):
    # 1. Extract 6-point Spectral Centroid sequence
    y, sr = librosa.load(file_path, duration=2.0)
    centroids = librosa.feature.spectral_centroid(y=y, sr=sr)[0]
    segments = np.array_split(centroids, 6)
    seq = [np.mean(seg) for seg in segments]
    
    # 2. Normalize and Plot
    norm_seq = np.array(seq) / np.max(seq)
    fig, ax = plt.subplots(figsize=(2, 2))
    ax.axhline(0, color='black', linewidth=1)
    
    colors = plt.cm.viridis(np.linspace(0, 0.8, len(norm_seq)-1))
    for i in range(len(norm_seq)-1):
        s, e = norm_seq[i], norm_seq[i+1]
        c, r = (s + e)/2, abs(e - s)/2
        if r < 0.001: continue
        
        # Draw circle
        circle = plt.Circle((c, 0), r, color=colors[i], fill=False, lw=1.5)
        ax.add_patch(circle)
        
        # Directions (both arrows pointing to end)
        d = 1 if e > s else -1
        for y_pos in [r, -r]:
            ax.annotate('', xy=(c + 0.05*d, y_pos), xytext=(c - 0.05*d, y_pos),
                        arrowprops=dict(arrowstyle='->', lw=1, color=colors[i]))
    
    ax.set_aspect('equal')
    plt.xlim(-0.2, 1.2); plt.ylim(-0.7, 0.7); plt.axis('off')
    
    # 3. Save as image for the model
    plt.savefig(save_path, dpi=64, bbox_inches='tight', pad_inches=0)
    plt.close()

# Run the transformation for all folders
for cls in classes:
    os.makedirs(os.path.join(output_img_path, cls), exist_ok=True)
    files = [f for f in os.listdir(os.path.join(data_path, cls)) if f.endswith('.wav')]
    for f in tqdm(files, desc=f"Processing {cls}"):
        generate_circular_image(os.path.join(data_path, cls, f), 
                               os.path.join(output_img_path, cls, f.replace('.wav', '.png')))
```
