```python
from mcap.reader import make_reader
from mcap_ros2.reader import read_ros2_messages

def get_mcap_topics(file_path):
    """
    Extracts all unique topics and their message types from an MCAP file.
    """
    topics_found = {}

    with open(file_path, "rb") as f:
        reader = make_reader(f)
        
        # Access the summary to get channel information
        # Channels contain the topic name and the schema (message type)
        summary = reader.get_summary()
        
        if summary is None:
            print("No summary found. The file might be unindexed.")
            return None

        for channel_id, channel in summary.channels.items():
            topic_name = channel.topic
            # Get the schema name (e.g., 'std_msgs/msg/String')
            schema = summary.schemas.get(channel.schema_id)
            schema_name = schema.name if schema else "Unknown Type"
            
            topics_found[topic_name] = schema_name

    return topics_found

if __name__ == "__main__":
    # Replace with your actual file path
    mcap_file = "/home/feliciano/Downloads/tank2_av_20260327_013503_816_0.mcap"
    
    try:
        results = get_mcap_topics(mcap_file)
        
        if results:
            print(f"{'TOPIC':<50} | {'TYPE'}")
            print("-" * 80)
            for topic, msg_type in sorted(results.items()):
                print(f"{topic:<50} | {msg_type}")
    except FileNotFoundError:
        print(f"Error: File '{mcap_file}' not found.")
    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
import numpy as np
import wave
from mcap_ros2.reader import read_ros2_messages

def mcap_to_wav(mcap_path, output_wav, target_topic="/tank2/hydrophone1/audio", sample_rate=48000):
    """
    Extracts Float32MultiArray data from MCAP and saves as a 44.1kHz WAV file.
    """
    audio_chunks = []

    print(f"Reading from {mcap_path}...")
    with open(mcap_path, "rb") as f:
        # Use the ROS 2 reader to decode the Float32MultiArray messages
        reader = read_ros2_messages(f, topics=[target_topic])
        
        for msg in reader:
            # Append the data attribute of the Float32MultiArray
            audio_chunks.append(np.array(msg.ros_msg.data, dtype=np.float32))

    if not audio_chunks:
        print("No audio data found on that topic.")
        return

    # 1. Combine all chunks into one continuous array
    full_signal = np.concatenate(audio_chunks)

    # 2. Normalize and Convert to 16-bit PCM (Short)
    # This prevents clipping and ensures the file is playable in standard players
    max_val = np.max(np.abs(full_signal))
    if max_val > 0:
        full_signal = full_signal / max_val  # Normalize to [-1, 1]
    
    # Scale to 16-bit integer range (-32768 to 32767)
    pcm_data = (full_signal * 32767).astype(np.int16)

    # 3. Write to WAV
    print(f"Saving to {output_wav} at {sample_rate}Hz...")
    with wave.open(output_wav, "wb") as wav_file:
        n_channels = 1  # Mono
        sampwidth = 2   # 2 bytes for 16-bit
        
        wav_file.setnchannels(n_channels)
        wav_file.setsampwidth(sampwidth)
        wav_file.setframerate(sample_rate)
        wav_file.writeframes(pcm_data.tobytes())

    print("Success!")

if __name__ == "__main__":
    # Update these paths as needed
    INPUT_MCAP = "/home/feliciano/Downloads/tank2_av_20260326_110514_075_0.mcap"
    OUTPUT_WAV = "Tank2_Hydro1.wav"
    
    try:
        mcap_to_wav(INPUT_MCAP, OUTPUT_WAV)
    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
from mcap.reader import make_reader
from mcap_ros2.reader import read_ros2_messages
import numpy as np

def analyze_mcap_audio(file_path, topic_name="/tank1/hydrophone2/audio"):
    timestamps = []
    samples_per_msg = []
    
    print(f"Analyzing file: {file_path}")
    print(f"Target topic: {topic_name}...")
    
    with open(file_path, "rb") as f:
        # read_ros2_messages returns an iterator of McapROS2Message objects
        reader = read_ros2_messages(f, topics=[topic_name])
        
        for msg in reader:
            # For McapROS2Message, the log time is available directly:
            timestamps.append(msg.log_time_ns)
            
            # Access the data inside the decoded ROS message
            samples_per_msg.append(len(msg.ros_msg.data))

    if not timestamps:
        print("No data found on the specified topic. Please check the topic name.")
        return

    # --- 1. Calculate Duration and Message Rate ---
    # Time difference between first and last message in seconds
    duration_ns = timestamps[-1] - timestamps[0]
    duration_sec = duration_ns / 1e9
    msg_count = len(timestamps)
    avg_msg_hz = msg_count / duration_sec if duration_sec > 0 else 0

    # --- 2. Calculate Effective Sampling Rate ---
    total_samples = sum(samples_per_msg)
    # Sampling Rate = Total Samples / Total Time in seconds
    calculated_sample_rate = total_samples / duration_sec if duration_sec > 0 else 0
    
    avg_samples_per_packet = np.mean(samples_per_msg)

    print("-" * 45)
    print(f"{'METRIC':<25} | {'VALUE'}")
    print("-" * 45)
    print(f"{'Total Duration':<25} | {duration_sec:.2f} seconds")
    print(f"{'Total Messages':<25} | {msg_count}")
    print(f"{'Avg Message Rate':<25} | {avg_msg_hz:.2f} Hz (packets/sec)")
    print(f"{'Avg Samples per Msg':<25} | {avg_samples_per_packet:.1f}")
    print(f"{'DETECTED SAMPLE RATE':<25} | {round(calculated_sample_rate)} Hz")
    print("-" * 45)
    
    return round(calculated_sample_rate)

if __name__ == "__main__":
    file_path = "/home/feliciano/Downloads/tank1_av_20260327_005039_640_0.mcap"
    try:
        detected_rate = analyze_mcap_audio(file_path)
    except FileNotFoundError:
        print("Error: Could not find the file at the path provided.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```

```python
import numpy as np
import wave
from mcap_ros2.reader import read_ros2_messages

def reconstruct_hydrophone_32bit_wav(mcap_path, output_wav, topic="/tank2/hydrophone1/audio", sample_rate=48000):
    """
    Reconstructs audio from MCAP Float32MultiArray into a 32-bit Float WAV file.
    """
    audio_segments = []

    print(f"Reading MCAP file: {mcap_path}")
    print(f"Extracting topic: {topic}")

    with open(mcap_path, "rb") as f:
        reader = read_ros2_messages(f, topics=[topic])
        
        for msg in reader:
            # Extract raw floats from the ROS message
            audio_segments.append(np.array(msg.ros_msg.data, dtype=np.float32))

    if not audio_segments:
        print("Error: No audio data found on that topic.")
        return

    # 1. Concatenate all message chunks
    full_signal = np.concatenate(audio_segments)
    
    # 2. Normalization (Optional but recommended for 32-bit Float)
    max_vol = np.max(np.abs(full_signal))
    if max_vol > 0:
        full_signal = full_signal / max_vol

    # 3. Write the 32-bit Float WAV file
    print(f"Writing 32-bit Float WAV to: {output_wav}...")
    with wave.open(output_wav, "wb") as wav_file:
        wav_file.setnchannels(1)      # Mono
        # 4 bytes per sample for 32-bit
        wav_file.setsampwidth(4)      
        wav_file.setframerate(sample_rate)
        
        # We must use .tobytes() on the float32 array
        wav_file.writeframes(full_signal.astype(np.float32).tobytes())

    print(f"Success! Exported {len(full_signal)} samples at 32-bit resolution.")

if __name__ == "__main__":
    INPUT_PATH = "/home/feliciano/Downloads/tank2_av_20260326_110514_075_0.mcap"
    OUTPUT_PATH = "tank2_hydrophone_32bit.wav"
    
    # Based on your analysis, 48000 is the logical hardware rate
    reconstruct_hydrophone_32bit_wav(INPUT_PATH, OUTPUT_PATH, sample_rate=48000)
```

```python
import numpy as np
import wave
from mcap_ros2.reader import read_ros2_messages
from scipy.signal import butter, lfilter, iirnotch

def butter_lowpass_filter(data, cutoff, fs, order=5):
    nyq = 0.5 * fs
    normal_cutoff = cutoff / nyq
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    return lfilter(b, a, data)

def apply_notch_filter(data, freq, fs, q=30):
    nyq = 0.5 * fs
    freq_norm = freq / nyq
    b, a = iirnotch(freq_norm, q)
    return lfilter(b, a, data)

def reconstruct_cleaned_audio(mcap_path, output_wav, topic="/tank2/hydrophone1/audio", sample_rate=48000):
    audio_segments = []

    print(f"Reading and cleaning MCAP: {mcap_path}")

    with open(mcap_path, "rb") as f:
        reader = read_ros2_messages(f, topics=[topic])
        
        for msg in reader:
            chunk = np.array(msg.ros_msg.data, dtype=np.float32)
            
            # --- STRATEGY 1: HEADER REMOVAL ---
            # If 1025 is the length, but audio is usually powers of 2 (1024), 
            # the first or last value might be a sequence ID or timestamp.
            if len(chunk) == 1025:
                chunk = chunk[1:] # Discard the first element
            
            audio_segments.append(chunk)

    if not audio_segments:
        print("No audio data found.")
        return

    # Combine
    full_signal = np.concatenate(audio_segments)

    # --- STRATEGY 2: LOW-PASS FILTER ---
    # Removes everything above 18kHz (where most "electronic noise" lives)
    print("Applying Low-Pass Filter (18kHz)...")
    full_signal = butter_lowpass_filter(full_signal, cutoff=18000, fs=sample_rate)

    # --- STRATEGY 3: NOTCH FILTER (Optional) ---
    # If there is a specific 'whine', you can uncomment this to kill a specific freq
    # full_signal = apply_notch_filter(full_signal, freq=15000, fs=sample_rate)

    # Normalize
    max_vol = np.max(np.abs(full_signal))
    if max_vol > 0:
        full_signal = full_signal / max_vol

    # Write 32-bit Float WAV
    with wave.open(output_wav, "wb") as wav_file:
        wav_file.setnchannels(1)
        wav_file.setsampwidth(4) # 32-bit
        wav_file.setframerate(sample_rate)
        wav_file.writeframes(full_signal.astype(np.float32).tobytes())

    print(f"Cleaned file saved to {output_wav}")

if __name__ == "__main__":
    INPUT_PATH = "/home/feliciano/Downloads/tank2_av_20260326_110514_075_0.mcap"
    OUTPUT_PATH = "tank2_hydrophone_cleaned_32bit.wav"
    
    # Run the reconstruction
    reconstruct_cleaned_audio(INPUT_PATH, OUTPUT_PATH)
```

```python
import numpy as np
import wave
from mcap_ros2.reader import read_ros2_messages

def reconstruct_optimized_audio(mcap_path, output_wav, topic="/tank2/hydrophone1/audio"):
    # We use 48000 as the hardware target based on the 1024/packet math
    NOMINAL_SAMPLE_RATE = 48000 
    audio_segments = []

    print(f"Processing: {mcap_path}")

    with open(mcap_path, "rb") as f:
        reader = read_ros2_messages(f, topics=[topic])
        
        for msg in reader:
            raw_data = np.array(msg.ros_msg.data, dtype=np.float32)
            
            # CRITICAL FIX: Strip the 1025th element metadata
            # We assume the first 1024 are audio and the 1025th is a header/footer
            # If the noise persists, try raw_data[1:] instead of raw_data[:1024]
            if len(raw_data) == 1025:
                clean_chunk = raw_data[:1024] 
            else:
                clean_chunk = raw_data
                
            audio_segments.append(clean_chunk)

    if not audio_segments:
        print("No data found.")
        return

    # Merge all cleaned chunks
    full_signal = np.concatenate(audio_segments)

    # REMOVE DC OFFSET: Hydrophones often have a constant voltage shift
    # This removes the "pop" at the start and center-aligns the wave
    full_signal -= np.mean(full_signal)

    # NORMALIZE: Scale to peak 1.0 for 32-bit Float standard
    max_val = np.max(np.abs(full_signal))
    if max_val > 0:
        full_signal = full_signal / max_val

    # WRITE 32-BIT FLOAT WAV
    print(f"Writing cleaned 32-bit WAV to: {output_wav}")
    with wave.open(output_wav, "wb") as wav_file:
        wav_file.setnchannels(1)      # Mono
        wav_file.setsampwidth(4)      # 32-bit (4 bytes)
        wav_file.setframerate(NOMINAL_SAMPLE_RATE)
        wav_file.writeframes(full_signal.astype(np.float32).tobytes())

    print("Reconstruction complete.")

if __name__ == "__main__":
    FILE = "/home/feliciano/Downloads/tank2_av_20260326_110514_075_0.mcap"
    OUT = "tank2_hydrophone_final_32bit.wav"
    reconstruct_optimized_audio(FILE, OUT)
```

```python
import numpy as np
import librosa
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from mcap_ros2.reader import read_ros2_messages
import warnings

# Suppress version warnings if any remain
warnings.filterwarnings("ignore", category=UserWarning)

def get_cleaned_audio(mcap_path, topic):
    segments = []
    with open(mcap_path, "rb") as f:
        reader = read_ros2_messages(f, topics=[topic])
        for msg in reader:
            chunk = np.array(msg.ros_msg.data, dtype=np.float32)
            # Apply our 1024-sample alignment fix
            if len(chunk) == 1025:
                segments.append(chunk[:1024])
            else:
                segments.append(chunk)
    return np.concatenate(segments) if segments else None

def cluster_hydrophone_signals(mcap_path, topic="/tank2/hydrophone1/audio", n_clusters=3):
    sr = 48000
    full_signal = get_cleaned_audio(mcap_path, topic)
    
    if full_signal is None:
        print("No data found.")
        return

    # PRE-PROCESSING: Simple high-frequency noise reduction
    # A 3-sample moving average acts as a very basic low-pass filter
    full_signal = np.convolve(full_signal, np.ones(3)/3, mode='same')

    # FEATURE EXTRACTION
    # We use a 1-second window to capture enough "texture" for the clusterer
    win_sec = 1.0
    hop_length = int(sr * win_sec)
    features = []
    times = []

    print("Analyzing frequency patterns...")
    for i in range(0, len(full_signal) - hop_length, hop_length):
        window = full_signal[i : i + hop_length]
        
        # MFCCs capture the "shape" of the sound spectrum
        mfccs = librosa.feature.mfcc(y=window, sr=sr, n_mfcc=13)
        features.append(np.mean(mfccs, axis=1))
        times.append(i / sr)

    # CLUSTERING
    X = StandardScaler().fit_transform(np.array(features))
    model = KMeans(n_clusters=n_clusters, n_init=10, random_state=42)
    labels = model.fit_predict(X)

    print("\n--- CLUSTER MAP ---")
    for c in range(n_clusters):
        indices = np.where(labels == c)[0]
        if len(indices) > 0:
            print(f"Cluster {c}: {len(indices)} occurrences. First at {times[indices[0]]:.2f}s")

    return times, labels, full_signal

if __name__ == "__main__":
    FILE = "/home/feliciano/Downloads/tank2_av_20260326_110514_075_0.mcap"
    try:
        t, clusters, audio = cluster_hydrophone_signals(FILE)
    except Exception as e:
        print(f"Update failed? Error: {e}")
```

```python
import numpy as np
import wave
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import librosa
from mcap_ros2.reader import read_ros2_messages

def save_cluster_wavs(mcap_path, topic="/tank2/hydrophone1/audio", n_clusters=3):
    sr = 48000
    audio_segments = []

    # 1. Extract and Clean Data
    print("Reading MCAP data...")
    with open(mcap_path, "rb") as f:
        reader = read_ros2_messages(f, topics=[topic])
        for msg in reader:
            chunk = np.array(msg.ros_msg.data, dtype=np.float32)
            # Alignment fix for the 1025-sample issue
            if len(chunk) == 1025:
                audio_segments.append(chunk[:1024])
            else:
                audio_segments.append(chunk)
    
    full_signal = np.concatenate(audio_segments)

    # 2. Feature Extraction for Clustering
    win_sec = 1.0
    hop_length = int(sr * win_sec)
    features = []
    chunk_indices = []

    print("Analyzing audio textures...")
    for i in range(0, len(full_signal) - hop_length, hop_length):
        window = full_signal[i : i + hop_length]
        mfccs = librosa.feature.mfcc(y=window, sr=sr, n_mfcc=13)
        features.append(np.mean(mfccs, axis=1))
        chunk_indices.append(i)

    # 3. Clustering
    X = StandardScaler().fit_transform(np.array(features))
    kmeans = KMeans(n_clusters=n_clusters, n_init=10, random_state=42)
    labels = kmeans.fit_predict(X)

    # 4. Export one WAV file per Cluster
    print("\nExporting files...")
    for cluster_id in range(n_clusters):
        # Find all audio indices belonging to this cluster
        cluster_indices = [chunk_indices[i] for i, label in enumerate(labels) if label == cluster_id]
        
        if not cluster_indices:
            continue

        # Stitch the matching audio chunks together
        cluster_audio = np.concatenate([full_signal[idx : idx + hop_length] for idx in cluster_indices])
        
        # Normalize and convert to 32-bit Float
        max_val = np.max(np.abs(cluster_audio))
        if max_val > 0:
            cluster_audio /= max_val

        output_filename = f"cluster_{cluster_id}_output.wav"
        with wave.open(output_filename, "wb") as wav_file:
            wav_file.setnchannels(1)
            wav_file.setsampwidth(4)  # 32-bit
            wav_file.setframerate(sr)
            wav_file.writeframes(cluster_audio.astype(np.float32).tobytes())
        
        print(f"Saved: {output_filename} ({len(cluster_audio)/sr:.2f} seconds)")

if __name__ == "__main__":
    FILE = "/home/feliciano/Downloads/tank2_av_20260326_110514_075_0.mcap"
    try:
        save_cluster_wavs(FILE)
    except Exception as e:
        print(f"Error: {e}\nEnsure you have downgraded to 'numpy<2' and restarted your kernel.")
```
