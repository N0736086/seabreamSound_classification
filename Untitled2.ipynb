```python
import os, torch, glob
import numpy as np
from scipy.io import wavfile
from scipy.signal import butter, lfilter
import torch.nn as nn

# 1. THE BLUEPRINT (Must be defined so PyTorch can load the weights)
class HR_PISFN_Segment(nn.Module):
    def __init__(self, n_input=1024, n_sums=64, n_classes=4):
        super().__init__()
        self.freq_bins = (n_input // 2) + 1
        self.register_buffer('window', torch.hann_window(n_input))
        self.spec_bn = nn.BatchNorm1d(self.freq_bins)
        self.spectral_weights = nn.Parameter(torch.randn(n_sums, self.freq_bins) * 0.01)
        self.stats_fc = nn.Sequential(nn.Linear(3, 16), nn.SiLU())
        self.classifier = nn.Sequential(
            nn.Linear(n_sums + 16, 64), nn.SiLU(), nn.Linear(64, n_classes)
        )

    def forward(self, x, s):
        mag = torch.abs(torch.fft.rfft(x.float() * self.window, dim=1))
        spec_feat = torch.matmul(self.spec_bn(mag), self.spectral_weights.t())
        return self.classifier(torch.cat((spec_feat, self.stats_fc(s.float())), dim=1))

# 2. THE CLASSIFICATION FUNCTION
def classify_with_names(file_path, model_path, root_data_path):
    # Detect classes exactly as they were during training
    class_names = sorted([d for d in os.listdir(root_data_path) if os.path.isdir(os.path.join(root_data_path, d))])
    n_classes = len(class_names)
    n_input = 1024

    # Load the Model
    model = HR_PISFN_Segment(n_classes=n_classes)
    
    # Load weights (using weights_only=True for security)
    if not os.path.exists(model_path):
        print(f"❌ Error: {model_path} not found. Run the training cell first.")
        return
        
    model.load_state_dict(torch.load(model_path, weights_only=True))
    model.eval()

    # Load and Filter Audio
    sr, data = wavfile.read(file_path)
    data = data.astype(np.float32)
    if np.max(np.abs(data)) > 1e-6: data /= np.max(np.abs(data))
    
    nyq = 0.5 * sr
    b, a = butter(5, 300/nyq, btype='high')
    data = lfilter(b, a, data).astype(np.float32)

    # Segmented Prediction (Slide across the whole file)
    predictions = []
    with torch.no_grad():
        for start in range(0, len(data) - n_input, n_input):
            chunk = data[start : start + n_input]
            norm = torch.tensor((chunk - np.mean(chunk)) / (np.std(chunk) + 1e-6)).unsqueeze(0)
            stats = torch.tensor([np.mean(chunk), np.std(chunk), np.max(chunk)]).unsqueeze(0)
            
            output = model(norm, stats)
            predictions.append(torch.argmax(output, dim=1).item())

    # Final Voting Result
    final_idx = max(set(predictions), key=predictions.count)
    predicted_folder = class_names[final_idx]
    
    # Identify actual folder from the file path provided
    # Works assuming standard /path/to/FOLDER/file.wav
    actual_folder = os.path.basename(os.path.dirname(file_path))
    consistency = (predictions.count(final_idx) / len(predictions)) * 100

    print(f"--- 📊 Long File Analysis ---")
    print(f"Target File:      {os.path.basename(file_path)}")
    print(f"Actual Folder:    {actual_folder}")
    print(f"Predicted Folder: {predicted_folder}")
    print(f"Match Status:     {'✅ CORRECT' if actual_folder == predicted_folder else '❌ INCORRECT'}")
    print(f"Consistency:      {consistency:.2f}%")
    
    if actual_folder != predicted_folder:
        print(f"\n💡 Note: The model is {consistency:.2f}% sure this is {predicted_folder}, "
              f"even though it came from {actual_folder}. This suggests a shared acoustic trait.")

# 3. RUN THE ANALYSIS
PATH = "/home/feliciano/FOURFISH"
MODEL_FILE = "lobster_perfect_model.pth"
TEST_FILE = "/home/feliciano/tester1.wav"

classify_with_names(TEST_FILE, MODEL_FILE, PATH)
```
