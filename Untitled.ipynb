```python
import os
import soundfile as sf

# Configuration
ROOT_DIR = "/home/feliciano/Downloads/Preliminary Data/AllFish/B2-hydro 5 fish"
OUTPUT_DIR = "/home/feliciano/Downloads/Preliminary Data/AllFish_2secSplit/B2-hydro 5 fish"
SEGMENT_SEC = 2.0

# Create the output directory if it doesn't exist
if not os.path.exists(OUTPUT_DIR):
    os.makedirs(OUTPUT_DIR)
    print(f"Created output directory: {OUTPUT_DIR}")

print(f"Starting 2-second segmentation in: {ROOT_DIR}\n")

# Process files directly in ROOT_DIR
for file in os.listdir(ROOT_DIR):
    # Skip if it's not a wav file
    if not file.lower().endswith(".wav"):
        continue

    wav_path = os.path.join(ROOT_DIR, file)

    # Read audio file
    audio, sr = sf.read(wav_path)
    
    # Convert to mono if stereo
    if audio.ndim > 1:
        audio = audio.mean(axis=1)

    segment_samples = int(SEGMENT_SEC * sr)
    total_samples = len(audio)

    # If file is shorter than 2 seconds, skip it or handle it
    if total_samples < segment_samples:
        print(f"  Skipping {file}: shorter than {SEGMENT_SEC} seconds.")
        continue

    base_name = os.path.splitext(file)[0]
    seg_idx = 0

    # Loop through the audio in steps of segment_samples
    for start in range(0, total_samples - segment_samples + 1, segment_samples):
        end = start + segment_samples
        segment = audio[start:end]

        start_t = start / sr
        end_t = end / sr

        # Format the output filename
        out_name = (
            f"{base_name}_seg{seg_idx:03d}_"
            f"{start_t:.1f}-{end_t:.1f}s.wav"
        )
        
        # Save to the OUTPUT_DIR
        out_path = os.path.join(OUTPUT_DIR, out_name)

        sf.write(out_path, segment, sr)
        seg_idx += 1

    print(f"  {file} → {seg_idx} segments saved.")

print(f"\nDone ✅ All files segmented and saved in: {OUTPUT_DIR}")
```
