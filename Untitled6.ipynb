```python
import numpy as np

class NeuralNetwork:
    def __init__(self, layers):
        """
        :param layers: List containing the number of neurons in each layer.
        e.g., [2, 4, 1] means 2 inputs, 4 hidden neurons, and 1 output.
        """
        self.layers = layers
        # Initialize weights and biases using He initialization for better convergence
        self.weights = [np.random.randn(layers[i], layers[i+1]) * np.sqrt(2.0/layers[i]) 
                        for i in range(len(layers)-1)]
        self.biases = [np.zeros((1, layers[i+1])) for i in range(len(layers)-1)]

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        return x * (1 - x)

    def forward_propagation(self, X):
        self.activations = [X]
        curr_input = X
        for i in range(len(self.weights)):
            # Net input: Z = XW + b
            z = np.dot(curr_input, self.weights[i]) + self.biases[i]
            # Activation: A = f(Z)
            curr_input = self.sigmoid(z)
            self.activations.append(curr_input)
        return self.activations[-1]

    def backward_propagation(self, X, y, learning_rate):
        m = y.shape[0]
        # Calculate output layer error
        error = self.activations[-1] - y
        delta = error * self.sigmoid_derivative(self.activations[-1])

        # Backpropagate through layers
        for i in reversed(range(len(self.weights))):
            # Gradient descent update
            grad_w = np.dot(self.activations[i].T, delta) / m
            grad_b = np.sum(delta, axis=0, keepdims=True) / m
            
            # Calculate delta for the next (previous) layer
            if i > 0:
                delta = np.dot(delta, self.weights[i].T) * self.sigmoid_derivative(self.activations[i])
            
            # Apply updates
            self.weights[i] -= learning_rate * grad_w
            self.biases[i] -= learning_rate * grad_b

    def train(self, X, y, epochs=10000, lr=0.1):
        for epoch in range(epochs):
            output = self.forward_propagation(X)
            self.backward_propagation(X, y, lr)
            if epoch % 1000 == 0:
                loss = np.mean(np.square(y - output)) # Mean Squared Error
                print(f"Epoch {epoch}, Loss: {loss:.6f}")

# --- Execution Example (XOR Problem) ---
if __name__ == "__main__":
    # Inputs (4 samples, 2 features)
    X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
    # Targets (XOR logic)
    y = np.array([[0], [1], [1], [0]])

    # Define Architecture: 2 Input -> 4 Hidden -> 1 Output
    nn = NeuralNetwork([2, 4, 1])
    
    print("Training the Neural Network...")
    nn.train(X, y, epochs=20000, lr=0.2)

    print("\nFinal Predictions:")
    final_output = nn.forward_propagation(X)
    for i in range(len(X)):
        print(f"Input: {X[i]} -> Predicted: {final_output[i][0]:.4f} (Actual: {y[i][0]})")
```
