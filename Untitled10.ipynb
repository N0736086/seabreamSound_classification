```python
import os
import numpy as np
import pandas as pd
import librosa
from collections import Counter
from tqdm import tqdm

# =========================
# sklearn
# =========================
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, cohen_kappa_score, matthews_corrcoef
)
from sklearn.svm import SVC
from sklearn.ensemble import (
    RandomForestClassifier, ExtraTreesClassifier,
    GradientBoostingClassifier, AdaBoostClassifier
)
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# =========================
# CONFIG
# =========================
DATASET_ROOT = "/home/feliciano/Downloads/Preliminary Data/AllFish_2secSplit"
SR = 22050
N_MFCC = 20
N_MELS = 64

# =========================
# FEATURE EXTRACTION
# =========================
def extract_features(wav_path):
    y, sr = librosa.load(wav_path, sr=SR)

    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=N_MFCC)
    mel = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=N_MELS)
    zcr = librosa.feature.zero_crossing_rate(y)
    centroid = librosa.feature.spectral_centroid(y=y, sr=sr)
    bandwidth = librosa.feature.spectral_bandwidth(y=y, sr=sr)
    rms = librosa.feature.rms(y=y)

    return np.hstack([
        mfcc.mean(axis=1), mfcc.std(axis=1),
        mel.mean(axis=1), mel.std(axis=1),
        zcr.mean(), zcr.std(),
        centroid.mean(), centroid.std(),
        bandwidth.mean(), bandwidth.std(),
        rms.mean(), rms.std()
    ])

# =========================
# LOAD DATA (GROUPED)
# =========================
X, y, groups = [], [], []

print("Extracting features...")

for fish_id in sorted(os.listdir(DATASET_ROOT)):
    fish_dir = os.path.join(DATASET_ROOT, fish_id)
    if not os.path.isdir(fish_dir):
        continue

    for file in os.listdir(fish_dir):
        if file.endswith(".wav"):
            feat = extract_features(os.path.join(fish_dir, file))
            X.append(feat)
            y.append(fish_id)
            groups.append(fish_id)

X = np.array(X)
y = np.array(y)
groups = np.array(groups)

le = LabelEncoder()
y_enc = le.fit_transform(y)
fish_labels = le.transform(np.unique(groups))

# =========================
# MODELS
# =========================
models = {
    "SVM-RBF": SVC(kernel="rbf"),
    "RandomForest": RandomForestClassifier(n_estimators=300),
    "ExtraTrees": ExtraTreesClassifier(n_estimators=300),
    "GradientBoosting": GradientBoostingClassifier(),
    "KNN": KNeighborsClassifier(n_neighbors=7),
    "LogisticRegression": LogisticRegression(max_iter=500),
    "NaiveBayes": GaussianNB(),
    "DecisionTree": DecisionTreeClassifier(),
    "AdaBoost": AdaBoostClassifier(),
    "LDA": LinearDiscriminantAnalysis()
}

# =========================
# LEAVE-ONE-FISH-OUT CV
# =========================
results = []

print("\nRunning Leave-One-Fish-Out CV...\n")

for name, model in models.items():
    true_fish = []
    pred_fish = []

    for test_fish in np.unique(groups):
        train_idx = groups != test_fish
        test_idx = groups == test_fish

        X_train, X_test = X[train_idx], X[test_idx]
        y_train = y_enc[train_idx]

        pipe = Pipeline([
            ("scaler", StandardScaler()),
            ("clf", model)
        ])

        pipe.fit(X_train, y_train)

        seg_preds = pipe.predict(X_test)

        # Majority vote
        voted = Counter(seg_preds).most_common(1)[0][0]

        true_fish.append(le.transform([test_fish])[0])
        pred_fish.append(voted)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(true_fish, pred_fish),
        "Precision_macro": precision_score(true_fish, pred_fish, average="macro"),
        "Recall_macro": recall_score(true_fish, pred_fish, average="macro"),
        "F1_macro": f1_score(true_fish, pred_fish, average="macro"),
        "Cohen_Kappa": cohen_kappa_score(true_fish, pred_fish),
        "MCC": matthews_corrcoef(true_fish, pred_fish)
    })

# =========================
# RESULTS
# =========================
df = pd.DataFrame(results).sort_values("Accuracy", ascending=False)

print("\n=== LEAKAGE-FREE RESULTS (FISH LEVEL) ===")
print(df)

# =========================
# CONFUSION MATRIX
# =========================
import seaborn as sns
import matplotlib.pyplot as plt

best_model = df.iloc[0]["Model"]
print(f"\nBest model: {best_model}")

cm = confusion_matrix(true_fish, pred_fish)

plt.figure(figsize=(10, 8))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=le.classes_,
    yticklabels=le.classes_
)
plt.xlabel("Predicted Fish")
plt.ylabel("True Fish")
plt.title("Leave-One-Fish-Out Confusion Matrix")
plt.tight_layout()
plt.show()
```
