```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import time
import os
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, f1_score
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, AdaBoostClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
import tensorflow as tf
from tensorflow.keras import layers, models

# --- 1. CONFIGURATION & DATA LOADING ---
FEATURE_DIR = Path("./FEATURES_MEL")
METADATA_PATH = "/home/feliciano/DATASETS/master_metadata_with_features.csv"

df = pd.read_csv(METADATA_PATH)
le = LabelEncoder()
df['label_int'] = le.fit_transform(df['behavior'])
num_classes = len(le.classes_)

def load_all_features(dataframe):
    x_2d, x_flat = [], []
    for path in dataframe['feature_path']:
        spec = np.load(FEATURE_DIR / path)
        # Normalize individual spectrogram
        spec = (spec - np.mean(spec)) / (np.std(spec) + 1e-6)
        x_2d.append(spec)
        x_flat.append(np.mean(spec, axis=1)) # Mean across time for ML
    return np.array(x_2d), np.array(x_flat)

print("Loading and preparing data...")
X_raw_2d, X_raw_flat = load_all_features(df)
y = df['label_int'].values

# Split Data
train_idx, test_idx = train_test_split(np.arange(len(y)), test_size=0.2, stratify=y, random_state=42)

# --- 2. DATA SHAPING FOR ALL MODEL TYPES ---
# A. ML & 1D-DL Data
X_train_flat, X_test_flat = X_raw_flat[train_idx], X_raw_flat[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

# B. Sequence Data (3D) for LSTM/GRU: (Samples, Timesteps, Features)
X_train_seq = X_train_flat.reshape(X_train_flat.shape[0], X_train_flat.shape[1], 1)
X_test_seq = X_test_flat.reshape(X_test_flat.shape[0], X_test_flat.shape[1], 1)

# C. 2D Data (4D) for PyTorch CNN: (Samples, 1, H, W)
X_train_2d = torch.tensor(X_raw_2d[train_idx], dtype=torch.float32).unsqueeze(1)
X_test_2d = torch.tensor(X_raw_2d[test_idx], dtype=torch.float32).unsqueeze(1)
y_train_torch = torch.tensor(y_train, dtype=torch.long)

# --- 3. THE 15-MODEL BENCHMARK ---
results = []

# --- CLASSICAL ML MODELS (Models 1-9) ---
ml_suite = {
    "Random Forest": RandomForestClassifier(),
    "Extra Trees": ExtraTreesClassifier(),
    "SVM (RBF)": SVC(probability=True),
    "XGBoost": XGBClassifier(),
    "KNN": KNeighborsClassifier(),
    "Logistic Reg": LogisticRegression(max_iter=1000),
    "Gaussian NB": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(),
    "AdaBoost": AdaBoostClassifier()
}

for name, model in ml_suite.items():
    start = time.time()
    model.fit(X_train_flat, y_train)
    preds = model.predict(X_test_flat)
    results.append({"Model": name, "Acc": accuracy_score(y_test, preds), "F1": f1_score(y_test, preds, average='weighted'), "Time": time.time()-start})

# --- 1D-DEEP LEARNING (Models 10-14) ---
def build_keras(model_type, input_shape):
    m = models.Sequential()
    m.add(layers.Input(shape=input_shape))
    if model_type == "LSTM": m.add(layers.LSTM(64))
    elif model_type == "GRU": m.add(layers.GRU(64))
    elif model_type == "1D-CNN": m.add(layers.Conv1D(64, 3, activation='relu')); m.add(layers.GlobalAveragePooling1D())
    elif model_type == "MLP": m.add(layers.Flatten()); m.add(layers.Dense(128, activation='relu'))
    m.add(layers.Dense(num_classes, activation='softmax'))
    return m

for name in ["LSTM", "GRU", "1D-CNN", "MLP"]:
    start = time.time()
    m = build_keras(name, (X_train_seq.shape[1], 1))
    m.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    m.fit(X_train_seq, y_train, epochs=10, verbose=0)
    preds = np.argmax(m.predict(X_test_seq), axis=1)
    results.append({"Model": name, "Acc": accuracy_score(y_test, preds), "F1": f1_score(y_test, preds, average='weighted'), "Time": time.time()-start})

# --- 2D-CNN PYTORCH (Model 15) ---
class FishCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(nn.Conv2d(1, 16, 3), nn.ReLU(), nn.MaxPool2d(2), nn.Flatten(), nn.Linear(16*63*15, num_classes)) # Adjust linear based on your shape
    def forward(self, x): return self.net(x)

# Note: Automatic shape inference for the Linear layer
def get_flatten_size(shape):
    m = nn.Sequential(nn.Conv2d(1, 16, 3), nn.ReLU(), nn.MaxPool2d(2), nn.Flatten())
    return m(torch.zeros(1, *shape)).shape[1]

class DynamicCNN(nn.Module):
    def __init__(self, in_shape):
        super().__init__()
        f_size = get_flatten_size(in_shape)
        self.conv = nn.Sequential(nn.Conv2d(1, 16, 3), nn.ReLU(), nn.MaxPool2d(2), nn.Flatten())
        self.fc = nn.Linear(f_size, num_classes)
    def forward(self, x): return self.fc(self.conv(x))

# Training PyTorch CNN
cnn = DynamicCNN(X_train_2d.shape[1:])
opt = torch.optim.Adam(cnn.parameters(), lr=0.001)
crit = nn.CrossEntropyLoss()
start = time.time()
for _ in range(10):
    opt.zero_grad(); crit(cnn(X_train_2d), y_train_torch).backward(); opt.step()
with torch.no_grad():
    cnn_preds = torch.max(cnn(X_test_2d), 1)[1].numpy()
results.append({"Model": "Proposed 2D-CNN", "Acc": accuracy_score(y_test, cnn_preds), "F1": f1_score(y_test, cnn_preds, average='weighted'), "Time": time.time()-start})

# --- 4. FINAL RESULTS ---
df_final = pd.DataFrame(results).sort_values(by="Acc", ascending=False)
print(df_final)
print("\n--- LATEX CODE ---")
print(df_final.to_latex(index=False))
```
