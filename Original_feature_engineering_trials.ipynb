```python
import matplotlib.pyplot as plt
import numpy as np

# Sequence from your sketch
path = [1, 3, 6, 4, 2, 5, 4, 2, 1,4, 7]

fig, ax = plt.subplots(figsize=(10, 6))

# Define the baseline number line (0 to 7)
x_axis = np.arange(0, 8)
ax.scatter(x_axis, np.zeros_like(x_axis), color='black', zorder=5)
for x in x_axis:
    ax.text(x, -0.5, str(x), ha='center', va='top', fontsize=12, fontweight='bold')

# Generate colors for each step
colors = plt.cm.plasma(np.linspace(0, 0.8, len(path)-1))

for i in range(len(path) - 1):
    start = path[i]
    end = path[i+1]
    
    center_x = (start + end) / 2
    radius = abs(end - start) / 2
    
    # Draw full circle based on the transition diameter
    circle = plt.Circle((center_x, 0), radius, color=colors[i], fill=False, linewidth=2, alpha=0.8)
    ax.add_patch(circle)
    
    # Directional indicator
    direction = 1 if end > start else -1
    ax.annotate('', xy=(end, 0), xytext=(end - 0.2*direction, 0.1),
                 arrowprops=dict(arrowstyle='->', color=colors[i], lw=2))

ax.set_aspect('equal')
plt.xlim(-1, 8)
plt.ylim(-3.5, 3.5)
plt.axis('off')
plt.title('Sequence Visualized as Full Circles')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sequence from your sketch
path = [1, 3, 6, 4, 2, 5, 0, 3, 1, 6]
colors = plt.cm.viridis(np.linspace(0, 0.8, len(path)-1))

fig, ax = plt.subplots(figsize=(12, 6))

# 1. Number line setup
x_axis = np.arange(0, 8)
ax.axhline(0, color='black', linewidth=1, zorder=1)
ax.scatter(x_axis, np.zeros_like(x_axis), color='black', zorder=5)
for x in x_axis:
    ax.text(x, -0.6, str(x), ha='center', va='top', fontsize=12, fontweight='bold')

# 2. Draw Circles with both arrows pointing toward the destination
for i in range(len(path) - 1):
    start, end = path[i], path[i+1]
    center_x = (start + end) / 2
    radius = abs(end - start) / 2
    
    # Draw full circle
    circle = plt.Circle((center_x, 0), radius, color=colors[i], fill=False, linewidth=2, alpha=0.7)
    ax.add_patch(circle)
    
    # Logic: Both arrows point toward 'end'. 
    # If end > start, both point Right (+). If end < start, both point Left (-).
    flow_direction = 1 if end > start else -1
    
    # Arrow on the top arc
    ax.annotate('', 
                xy=(center_x + 0.1 * flow_direction, radius), 
                xytext=(center_x - 0.1 * flow_direction, radius),
                arrowprops=dict(arrowstyle='->', color=colors[i], lw=2.5, mutation_scale=20))
    
    # Arrow on the bottom arc (pointing the SAME direction as the top)
    ax.annotate('', 
                xy=(center_x + 0.1 * flow_direction, -radius), 
                xytext=(center_x - 0.1 * flow_direction, -radius),
                arrowprops=dict(arrowstyle='->', color=colors[i], lw=2.5, mutation_scale=20))

ax.set_aspect('equal')
plt.xlim(-1, 8)
plt.ylim(-4, 4)
plt.axis('off')
plt.title(f"Sequence: {' -> '.join(map(str, path))} (Convergent Flow)", fontsize=14)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Original Sequence and Max Normalization
sequence = np.array([1, 3, 6, 4, 2, 5])
norm_seq = sequence / sequence.max()

# 2. Setup the Plot (focusing purely on the result plot)
fig, ax = plt.subplots(figsize=(12, 6))
colors = plt.cm.viridis(np.linspace(0, 0.8, len(norm_seq)-1))

# Draw the main horizontal line for the sequence path
ax.axhline(0, color='black', linewidth=1.5, zorder=1)

# Draw the normalized points (nodes) without numerical labels
ax.scatter(norm_seq, np.zeros_like(norm_seq), color='black', s=60, zorder=5)

# 3. Draw the Circles and Convergent Arrows
for i in range(len(norm_seq) - 1):
    start, end = norm_seq[i], norm_seq[i+1]
    center_x = (start + end) / 2
    radius = abs(end - start) / 2
    
    # Draw the full circle
    circle = plt.Circle((center_x, 0), radius, color=colors[i], fill=False, linewidth=2.5, alpha=0.8)
    ax.add_patch(circle)
    
    # Determine direction (Both arrows point to the 'end' point)
    flow_dir = 1 if end > start else -1
    
    # Top arc arrow
    ax.annotate('', xy=(center_x + 0.02 * flow_dir, radius), xytext=(center_x - 0.02 * flow_dir, radius),
                arrowprops=dict(arrowstyle='->', color=colors[i], lw=2.5, mutation_scale=20))
    
    # Bottom arc arrow
    ax.annotate('', xy=(center_x + 0.02 * flow_dir, -radius), xytext=(center_x - 0.02 * flow_dir, -radius),
                arrowprops=dict(arrowstyle='->', color=colors[i], lw=2.5, mutation_scale=20))

# 4. Final Cleanup: Remove all numbers, axes, and background clutter
ax.set_aspect('equal')
plt.xlim(norm_seq.min() - 0.1, norm_seq.max() + 0.1)
plt.ylim(-0.6, 0.6)
plt.axis('off') # This removes all numbers and axis lines

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import os

# Create directory for synthetic images
os.makedirs("synthetic_dataset", exist_ok=True)

def plot_circular_sequence(seq, class_name, filename):
    # Max Normalization (Divide by highest value)
    norm_seq = np.array(seq) / np.max(seq)
    
    fig, ax = plt.subplots(figsize=(6, 4))
    colors = plt.cm.viridis(np.linspace(0, 0.8, len(norm_seq)-1))
    
    # Base path line
    ax.axhline(0, color='black', linewidth=1.5, zorder=1)
    ax.scatter(norm_seq, np.zeros_like(norm_seq), color='black', s=40, zorder=5)

    for i in range(len(norm_seq) - 1):
        start, end = norm_seq[i], norm_seq[i+1]
        center_x = (start + end) / 2
        radius = abs(end - start) / 2
        
        # Avoid zero-radius circles if numbers repeat
        if radius == 0: continue
            
        circle = plt.Circle((center_x, 0), radius, color=colors[i], fill=False, linewidth=2)
        ax.add_patch(circle)
        
        # Direction logic: Both arrows point to 'end'
        flow_dir = 1 if end > start else -1
        
        # Top and Bottom Arrows
        for y_pos in [radius, -radius]:
            ax.annotate('', xy=(center_x + 0.02 * flow_dir, y_pos), 
                        xytext=(center_x - 0.02 * flow_dir, y_pos),
                        arrowprops=dict(arrowstyle='->', color=colors[i], lw=1.5, mutation_scale=10))

    ax.set_aspect('equal')
    plt.xlim(-0.1, 1.1)
    plt.ylim(-0.6, 0.6)
    plt.axis('off')
    plt.savefig(f"synthetic_dataset/{filename}.png", dpi=100, bbox_inches='tight')
    plt.close()

# Define Synthetic Sequences
synthetic_data = {
    "Ascending": [1, 2, 4, 3, 5, 6],
    "Descending": [6, 5, 2, 4, 1, 3],
    "Oscillating": [1, 6, 2, 5, 1, 6],
    "Random": [3, 1, 6, 2, 5, 4]
}

# Generate and Save
for name, seq in synthetic_data.items():
    print(f"Generating plot for {name} class...")
    plot_circular_sequence(seq, name, f"sample_{name}")

print("\nDone! Check the 'synthetic_dataset' folder for your 4 class samples.")
```
