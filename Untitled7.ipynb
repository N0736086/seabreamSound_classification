```python
import os
import numpy as np
import noisereduce as nr
from moviepy import VideoFileClip, AudioArrayClip

def remove_audio_noise(input_video_path, output_video_path):
    # 1. Load the video file
    print(f"Loading video: {input_video_path}")
    video = VideoFileClip(input_video_path)
    
    # 2. Extract the audio and convert to a numpy array
    # Standard sample rate for high-quality audio
    sr = 44100
    audio_data = video.audio.to_soundarray(fps=sr)
    
    print("Processing audio noise reduction...")
    
    # 3. Apply noise reduction
    # y.T transposes the array to (channels, samples) as required by noisereduce
    reduced_noise = nr.reduce_noise(y=audio_data.T, sr=sr, stationary=True)
    
    # 4. Create a new AudioArrayClip from the cleaned data
    # Transpose back to (samples, channels) for MoviePy
    clean_audio = AudioArrayClip(reduced_noise.T, fps=sr)
    
    # 5. Set the cleaned audio back to the original video
    # In v2.x, .with_audio() is used instead of .set_audio()
    final_video = video.with_audio(clean_audio)
    
    # 6. Write the result to the directory
    # Fixed: Passing the path as the first positional argument 'filename'
    print(f"Saving cleaned video to: {output_video_path}")
    final_video.write_videofile(
        output_video_path, 
        codec="libx264", 
        audio_codec="aac",
        temp_audiofile='temp-audio.m4a',
        remove_temp=True
    )
    
    # Cleanup to release file locks
    video.close()
    final_video.close()

if __name__ == "__main__":
    # Ensure the directory exists
    output_dir = "python"
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
        
    input_file = "/home/feliciano/Videos/Webcam/2026-01-27-153152.webm"
    output_file = os.path.join(output_dir, "cleaned_video.mp4")
    
    if os.path.exists(input_file):
        try:
            remove_audio_noise(input_file, output_file)
            print("Successfully cleaned video!")
        except Exception as e:
            print(f"An error occurred: {e}")
    else:
        print(f"Error: {input_file} not found.")
```
