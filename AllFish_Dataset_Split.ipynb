```python
import os
import soundfile as sf

ROOT_DIR = "/home/feliciano/Downloads/Preliminary Data/AllFish"
SEGMENT_SEC = 2.0

print("Starting 2-second segmentation...\n")

for subfolder in os.listdir(ROOT_DIR):
    subfolder_path = os.path.join(ROOT_DIR, subfolder)

    if not os.path.isdir(subfolder_path):
        continue

    print(f"Processing folder: {subfolder}")

    for file in os.listdir(subfolder_path):
        if not file.lower().endswith(".wav"):
            continue

        wav_path = os.path.join(subfolder_path, file)

        audio, sr = sf.read(wav_path)
        if audio.ndim > 1:
            audio = audio.mean(axis=1)  # convert to mono

        segment_samples = int(SEGMENT_SEC * sr)
        total_samples = len(audio)

        base_name = os.path.splitext(file)[0]
        seg_idx = 0

        for start in range(0, total_samples - segment_samples + 1, segment_samples):
            end = start + segment_samples
            segment = audio[start:end]

            start_t = start / sr
            end_t = end / sr

            out_name = (
                f"{base_name}_seg{seg_idx:03d}_"
                f"{start_t:.1f}-{end_t:.1f}s.wav"
            )
            out_path = os.path.join(subfolder_path, out_name)

            sf.write(out_path, segment, sr)
            seg_idx += 1

        print(f"  {file} → {seg_idx} segments")

print("\nDone ✅ All files segmented.")
```

```python
import os
import shutil

ROOT_DIR = "/home/feliciano/Downloads/Preliminary Data/AllFish"
OUT_DIR = ROOT_DIR + "_2secSplit"

os.makedirs(OUT_DIR, exist_ok=True)

for subfolder in os.listdir(ROOT_DIR):
    subfolder_path = os.path.join(ROOT_DIR, subfolder)

    if not os.path.isdir(subfolder_path):
        continue

    # Create corresponding folder in OUT_DIR
    out_subfolder = os.path.join(OUT_DIR, subfolder)
    os.makedirs(out_subfolder, exist_ok=True)

    for file in os.listdir(subfolder_path):
        # Identify the segmented files
        if "_seg" in file and file.lower().endswith(".wav"):
            src = os.path.join(subfolder_path, file)
            dst = os.path.join(out_subfolder, file)

            shutil.copy2(src, dst)

print("Done ✅ Segmented files copied into the new folder.")
```
