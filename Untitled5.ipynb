```python
import os
import librosa
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import classification_report

# 1. SETUP PATHS
# Adjust 'main_folder' to the path where your 5 folders are located
main_folder = '/home/feliciano/Downloads/Preliminary Data/AllFish_2secSplit'
classes = os.listdir(main_folder) # Assumes folders are named by class

def extract_sequence_from_audio(file_path, n_segments=6):
    """Extracts a sequence of 6 values from a 2-second audio clip."""
    try:
        y, sr = librosa.load(file_path, duration=2.0)
        # We use Spectral Centroid as the 'value' for our sequence
        # We split the 2s audio into n_segments and take the mean frequency of each
        spectral_centroids = librosa.feature.spectral_centroid(y=y, sr=sr)[0]
        
        # Split centroids into 6 equal parts and take the mean of each
        segments = np.array_split(spectral_centroids, n_segments)
        sequence = [np.mean(seg) for seg in segments]
        return sequence
    except Exception as e:
        print(f"Error processing {file_path}: {e}")
        return None

# 2. LOAD DATASET
X_audio_raw = []
y_labels = []

print("Extracting features from audio folders...")
for label_idx, class_name in enumerate(classes):
    class_path = os.path.join(main_folder, class_name)
    if not os.path.isdir(class_path): continue
    
    for file_name in os.listdir(class_path):
        if file_name.endswith(('.wav', '.mp3')):
            file_path = os.path.join(class_path, file_name)
            seq = extract_sequence_from_audio(file_path)
            if seq:
                X_audio_raw.append(seq)
                y_labels.append(label_idx)

# 3. CIRCULAR TRANSFORMATION (Using the same logic from your drawing)
def transform_to_circles(X_data):
    X_numeric = np.array(X_data).astype(float)
    circular_data = []
    for seq in X_numeric:
        max_val = np.max(seq)
        norm_seq = seq / max_val if max_val != 0 else seq
        
        diffs = np.diff(norm_seq)
        diameters = np.abs(diffs)
        directions = np.sign(diffs)
        centers = (norm_seq[:-1] + norm_seq[1:]) / 2
        
        feat_vector = np.hstack([diameters, directions, centers])
        circular_data.append(feat_vector)
    return np.array(circular_data)

X_circular = transform_to_circles(X_audio_raw)
y_array = np.array(y_labels)

# 4. TRAIN/TEST/VAL SPLIT (70/20/10)
X_train, X_tmp, y_train, y_tmp = train_test_split(
    X_circular, y_array, test_size=0.30, stratify=y_array, random_state=42
)
X_test, X_val, y_test, y_val = train_test_split(
    X_tmp, y_tmp, test_size=1/3, stratify=y_tmp, random_state=42
)

# 5. MODEL TRAINING
clf = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(kernel='rbf', C=1.0, probability=True))
])

clf.fit(X_train, y_train)

# 6. RESULTS
print("\n--- Real Audio Circular Classification Report ---")
print(classification_report(y_test, clf.predict(X_test), target_names=classes))
```
