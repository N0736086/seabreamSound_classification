```python
import os
import soundfile as sf
import numpy as np

# =========================
# USER PARAMETERS
# =========================
INPUT_ROOT = "/home/feliciano/Portugal_FishSounds/DATASET_SEABREAM/Dataset_Research_Batch/REAL RECORDINGS_ALGARVE_JULY_2025/INDIVIDUAL FISH CLASS"
OUTPUT_ROOT = "SEGMENTED_2S_1S_HOP"

WINDOW_SEC = 2.0   # segment length
HOP_SEC = 1.0      # slide step

# =========================
# PROCESSING
# =========================
os.makedirs(OUTPUT_ROOT, exist_ok=True)

for root, _, files in os.walk(INPUT_ROOT):
    for file in files:
        if not file.lower().endswith(".wav"):
            continue

        wav_path = os.path.join(root, file)

        # Read audio
        audio, sr = sf.read(wav_path)
        if audio.ndim > 1:
            audio = audio.mean(axis=1)  # convert to mono if needed

        window_samples = int(WINDOW_SEC * sr)
        hop_samples = int(HOP_SEC * sr)

        total_samples = len(audio)

        # Relative path to preserve folder structure
        rel_dir = os.path.relpath(root, INPUT_ROOT)
        out_dir = os.path.join(OUTPUT_ROOT, rel_dir)
        os.makedirs(out_dir, exist_ok=True)

        base_name = os.path.splitext(file)[0]

        seg_idx = 0
        for start in range(0, total_samples - window_samples + 1, hop_samples):
            end = start + window_samples
            segment = audio[start:end]

            start_time = start / sr
            end_time = end / sr

            out_name = (
                f"{base_name}_seg{seg_idx:04d}_"
                f"{start_time:.2f}-{end_time:.2f}s.wav"
            )
            out_path = os.path.join(out_dir, out_name)

            sf.write(out_path, segment, sr)
            seg_idx += 1

        print(f"Processed: {wav_path} → {seg_idx} segments")
```
