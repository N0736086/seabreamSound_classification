```python
import os
import librosa
import soundfile as sf

# ---------------------------------------------------------
# CONFIGURATION
# ---------------------------------------------------------
INPUT_BASE = "/home/feliciano/Documents/DATASET_SEABREAM"
OUTPUT_BASE = "/home/feliciano/Documents/DATASET_SEABREAM_SEGMENTED"

#SEGMENT_DURATIONS = [1, 2, 5, 10, 50]  # seconds
SEGMENT_DURATIONS = [20]  # seconds

AUDIO_EXT = (".wav", ".WAV")

# ---------------------------------------------------------
# FUNCTION TO SEGMENT A SINGLE FILE
# ---------------------------------------------------------
def segment_audio_file(input_path, output_root, segment_seconds):
    audio, sr = librosa.load(input_path, sr=None, mono=True)
    total_samples = len(audio)
    segment_samples = segment_seconds * sr

    # Compute number of segments
    num_segments = total_samples // segment_samples

    # Build output directory
    rel_path = os.path.relpath(input_path, INPUT_BASE)
    rel_dir = os.path.dirname(rel_path)
    output_dir = os.path.join(output_root, rel_dir)
    os.makedirs(output_dir, exist_ok=True)

    base_name = os.path.splitext(os.path.basename(input_path))[0]

    for i in range(num_segments):
        start = i * segment_samples
        end = start + segment_samples
        segment = audio[start:end]

        out_name = f"{base_name}_seg{str(i+1).zfill(4)}.wav"
        out_path = os.path.join(output_dir, out_name)

        sf.write(out_path, segment, sr)

    print(f"Segmented {input_path} into {num_segments} chunks of {segment_seconds}s.")


# ---------------------------------------------------------
# MAIN LOOP
# ---------------------------------------------------------
def main():
    for root, dirs, files in os.walk(INPUT_BASE):
        for file in files:
            if file.endswith(AUDIO_EXT):
                input_path = os.path.join(root, file)

                for seg_dur in SEGMENT_DURATIONS:
                    output_root = os.path.join(OUTPUT_BASE, f"Segmented_{seg_dur}s")
                    segment_audio_file(input_path, output_root, seg_dur)


if __name__ == "__main__":
    main()
```
